# **Feature Engineering**

```python
df['Year'] = df['Date'].dt.year
```

```python
df['is_six'] = (df['Batter Runs'] == 6).astype(int)
```

```python
df['Method'].unique()
```

```python
valid_method = ['bowled', 'caught', 'lbw', 'stumped','caught and bowled','hit wicket']

df['bowler_wickets'] = df['Method'].isin(valid_method)
```

```python
df['is_dot_balls'] = ((df['Runs From Ball'] == 0) & (df['Valid Ball'] == 1)).astype(int)
```

# **EDA & Visualization**

## Orange Cap Winners

### EDA

```python
batters_yrly_runs = df.groupby(['Year', 'Batter'])['Batter Runs'].sum().reset_index()

orange_cap_indx = batters_yrly_runs.groupby('Year')['Batter Runs'].idxmax()

orange_cap_winners = batters_yrly_runs.loc[orange_cap_indx]

orange_cap_winners = orange_cap_winners.rename(columns={"Batter Runs": "Total Batter Runs"})
orange_cap_winners
```

### Visualize

```python
plt.figure(figsize=(12, 7))
bars = plt.barh(
    orange_cap_winners['Year'],
    orange_cap_winners['Total Batter Runs'],
    alpha=0.7
)

# adding data to bars
for year, batter, runs in zip(
    orange_cap_winners['Year'],
    orange_cap_winners['Batter'],
    orange_cap_winners['Total Batter Runs']
):

    # Batter name inside bar
    plt.text(
        runs * 0.02,
        year,
        batter,
        ha='left',
        va='center',
        fontsize=10,
        fontweight='bold'
    )

    # Runs at end of bar
    plt.text(
        runs + 10,
        year,
        str(runs),
        ha='left',
        va='center',
        fontsize=10,
        fontweight='bold'
    )

plt.xlabel('Batter Runs', fontsize=13)
plt.ylabel('IPL Years', fontsize=13)
plt.title('Orange Cap Winners', fontsize=16)

plt.xlim(0, orange_cap_winners['Total Batter Runs'].max() + 150)
plt.tight_layout()
plt.show()
```

## Batter With Highest Sixes

### EDA

```python
total_batters_sixes = df.groupby(['Year', 'Batter'])['is_six'].sum().reset_index()
highest_sixes_scorer_idx = total_batters_sixes.groupby('Year')['is_six'].idxmax()
highest_sixes_scorer = total_batters_sixes.loc[highest_sixes_scorer_idx]

highest_sixes_scorer = highest_sixes_scorer.rename(columns={"is_six": "Total Sixes"})
highest_sixes_scorer
```

### Visualize

```python
plt.figure(figsize=(12, 7))

bars = plt.barh(
    highest_sixes_scorer['Year'].astype(str),
    highest_sixes_scorer['Total Sixes'],
    alpha=0.7
)

for idx, (year, batter, sixes) in enumerate(
    zip(
        highest_sixes_scorer['Year'],
        highest_sixes_scorer['Batter'],
        highest_sixes_scorer['Total Sixes']
    )
):
  plt.text(sixes * 0.02, idx, batter, ha='left', va='center', fontsize=10, fontweight='bold')
  plt.text(sixes +1, idx, sixes, ha='left', va='center', fontsize=10, fontweight='bold')

plt.xlabel('IPL Year')
plt.ylabel('Total Sixes')
plt.title('Batsman With Highest Number of Sixes')

plt.xlim(0, highest_sixes_scorer['Total Sixes'].max() + 50)
plt.tight_layout()
plt.show()
```

## Purple Cap Winners(Bowlers with Highest Wicket takers in each Season)

### EDA

```python
# Season wise wicket takers
yrly_wicket_takers = df.groupby(['Year', 'Bowler'])['bowler_wickets'].sum().reset_index()

# Bowler with highest wicket takings index
highest_wckt_takers_idx = yrly_wicket_takers.groupby('Year')['bowler_wickets'].idxmax()

# Highest Wicket takers details each Season
highest_wicket_takers = yrly_wicket_takers.loc[highest_wckt_takers_idx]

highest_wicket_takers = highest_wicket_takers.rename(columns = {'bowler_wickets':'Total Wickets Taken'})
```

### Visualize

```python
# Plot figure and chart
plt.figure(figsize=(14, 7))
bars = plt.bar(
          highest_wicket_takers['Year'].astype(str),
          highest_wicket_takers['Total Wickets Taken'],
          color='royalblue',
          alpha=0.7
      )

# Adding names of bowlers with their wickets
for bar, wickets, bowler in zip(bars, highest_wicket_takers['Total Wickets Taken'], highest_wicket_takers['Bowler']):
  plt.text(
      bar.get_x() + bar.get_width() / 2,
      bar.get_height() + 2,
      f'{bowler} \n {wickets}',
      ha='center', va='center', fontsize=10, fontweight='bold'
  )

# visualize with labels and titles
plt.xlabel('IPL Year', fontsize=13)
plt.ylabel('Total Wickets', fontsize=13)
plt.title('Purple Cap Winners(Highest Wicket takers in each Season)', fontsize=16)

# styling visualization
plt.ylim(0, highest_wicket_takers['Total Wickets Taken'].max() + 10)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Bowlers with Highest Dot Balls

### EDA

```python
# Get total dot balls
bowlers_with_dot_balls = df.groupby(['Year', 'Bowler'])['is_dot_balls'].sum().reset_index()

# Get index of highest dot balls in each season
highest_dot_bowlers_yrly_idx = bowlers_with_dot_balls.groupby('Year')['is_dot_balls'].idxmax()

# Get season wise highest dot bowlers data
highest_dot_bowlers_yrly = bowlers_with_dot_balls.loc[highest_dot_bowlers_yrly_idx]

# Changind column name
highest_dot_bowlers_yrly = highest_dot_bowlers_yrly.rename(columns={'is_dot_balls': 'Total Dot Balls'})
```

### Visualization

```python
# Visualization plot figure
plt.figure(figsize=(14, 7))
bars = plt.bar(
    highest_dot_bowlers_yrly['Year'].astype(str),
    highest_dot_bowlers_yrly['Total Dot Balls'],
    color='royalblue',
    alpha=0.7
)

# Naming each bars
for bar, bowler, dots in zip(
    bars,
    highest_dot_bowlers_yrly['Bowler'],
    highest_dot_bowlers_yrly['Total Dot Balls']
):
  plt.text(
      bar.get_x() + bar.get_width() / 2,
      bar.get_height() + 10,
      f'{bowler}\n{dots}',
      ha='center', va='center', fontsize=10, fontweight='bold'
  )

# Adding labels and title
plt.xlabel('IPL Year(Season)', fontsize=13)
plt.ylabel('Total Dot Balls', fontsize=13)
plt.title('Bowlers with Highest Dot Balls', fontsize=16, fontweight='bold')

# Designing the chart
plt.ylim(0, highest_dot_bowlers_yrly['Total Dot Balls'].max() + 50)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Highest Target Score set in each Season

### Using Matplotlib Line chart

```python
plt.figure(figsize=(14, 7))

# Get highest target score set
highest_target_set_each_season = df.groupby('Year')['Target Score'].max().reset_index()

plt.plot(
    highest_target_set_each_season['Year'],
    highest_target_set_each_season['Target Score'],
    color='green', marker='o', alpha=0.7
)

plt.xlabel('Seasons(Years)', fontsize=13)
plt.ylabel('Target Score', fontsize=13)
plt.title('Highest Target Score for each IPL Season(Years)', fontsize=16, fontweight='bold')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

## Lowest Target Score for each Season

```python
plt.figure(figsize=(14, 7))

minimum_target_score = df.groupby('Year')['Target Score'].min().reset_index()

sns.lineplot(data= minimum_target_score, x='Year', y='Target Score', marker='o', color='green')

plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.xlabel('IPL Season(Years)', fontsize=13)
plt.ylabel('Targest Scores', fontsize=13)
plt.title('Lowest Target Scores each Season', fontsize=16, fontweight='bold')

plt.show()
```
